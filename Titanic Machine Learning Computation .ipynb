```python
def view_ser(data):
    print(data.head())
def data_type(data):
    print(data.dtypes)
def shape_sur(data):
    print(f'Row : {data.shape[0]} , Column : {data.shape[1]}')
def handle_missing(data):
    missing = data.isnull().sum()
    missing_per = missing / len(data) * 100
    result_data = pd.DataFrame({'Missing_Count': missing,'Missing_%' : missing_per})
    return result_data[result_data['Missing_Count']>0 ].sort_values(by='Missing_Count', ascending=False)
def check_traget(data,target_col):
    print(data[target_col].value_counts(normalize=True).round(2))
def dup_count(data):
    duplicate_col = data.duplicated().sum()
    print(f'Duplicate Columns : {duplicate_col}')
def all_data(data,target_col=None):
    print('='* 50)
    view_ser(data)
    print('=== DATA TYPE===')
    data_type(data)
    print('=== SHAPE ===')
    shape_sur(data)
    print('=== MISSING VALUES ===')
    print(handle_missing(data))
    print('===DUPLICATES===')
    dup_count(data)
    if check_traget:
        print(f'==={target_col}===')
        check_traget(data,target_col)
```

## 3) Handling missing values
1. === MISSING VALUES ===
          Missing_Count  Missing_%
Cabin               687  77.104377
Age                 177  19.865320
Embarked              2   0.224467

2. need to handle the missing record's in the 

```python
def handling_missing(data,col, strategy ='mean'):
    if strategy == 'mean':
        data[col] = data[col].fillna(data[col].mean()).round(1)
    elif strategy == 'median':
        data[col] = data[col].fillna(data[col].median())
    elif strategy == 'mode':
        data[col] = data[col].fillna(data[col].mode()[0])
    elif strategy == 'drop_col':
        data[col] = data.drop(columns = [col])
    elif strategy == 'drop_row' : 
        data[col] = data.dropna(subset=[col])
    return data
    
```

Hadling the missing value 'AGE'

```python
handling_missing(test_data,'Age','mean')
```

```python
handling_missing(test_data,'Cabin','mode')
```

```python
handling_missing(test_data, 'Embarked', 'mode')
```
