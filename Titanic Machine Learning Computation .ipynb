## 3) Handling missing values
1. === MISSING VALUES ===
          Missing_Count  Missing_%
Cabin               687  77.104377
Age                 177  19.865320
Embarked              2   0.224467

2. need to handle the missing record's in the 

```python
def handling_missing(data,col, strategy ='mean'):
    if strategy == 'mean':
        data[col] = data[col].fillna(data[col].mean()).round(1)
    elif strategy == 'median':
        data[col] = data[col].fillna(data[col].median())
    elif strategy == 'mode':
        data[col] = data[col].fillna(data[col].mode()[0])
    elif strategy == 'drop_col':
        data[col] = data.drop(columns = [col])
    elif strategy == 'drop_row' : 
        data[col] = data.dropna(subset=[col])
    return data
    
```

## 4) Identify or Handling Outer layer
let understand the how our data is far from rest of the data
* [IQR = Q3 -Q1](http://)
* [Q1 = q1 - 1.5 (IQR)](http://)
* [Q3 = q3 + 1.5 (IQR)](http://)
* Babies/toddlers (0.8, 1, 2 years old)
* Older adults (55–74 years old)

```python
def identif_outlayer(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 *IQR
    upper = Q3 + 1.5 *IQR
    outlayer = data[(data[col] < lower) | (data[col] > upper)]
    #outlayer = data[(data[col] < lower) | (data[col] > upper)]
    print(f'{col} : {len(outlayer)} outlayer (outside{lower:.1f} to {upper:.1f})')
    #print(f'{col}: {len(outlayer)} outlayer (outside {lower:.1f} to {upper:.1f})')
    return outlayer
#outlayer_rows = identif_outlayer(train_data, 'Age')
#print(outlayer_rows[['Age', 'Survived', 'Pclass']])
```
