## 2) Understand and explor the dataset

```python
def view_ser(data):
    print(data.head())
def data_type(data):
    print(data.dtypes)
def shape_sur(data):
    print(f'Row : {data.shape[0]} , Column : {data.shape[1]}')
def handle_missing(data):
    missing = data.isnull().sum()
    missing_per = missing / len(data) * 100
    result_data = pd.DataFrame({'Missing_Count': missing,'Missing_%' : missing_per})
    return result_data[result_data['Missing_Count']>0 ].sort_values(by='Missing_Count', ascending=False)
def check_traget(data,target_col):
    print(data[target_col].value_counts(normalize=True).round(2))
def dup_count(data):
    duplicate_col = data.duplicated().sum()
    print(f'Duplicate Columns : {duplicate_col}')
def all_data(data,target_col=None):
    print('='* 50)
    view_ser(data)
    print('=== DATA TYPE===')
    data_type(data)
    print('=== SHAPE ===')
    shape_sur(data)
    print('=== MISSING VALUES ===')
    print(handle_missing(data))
    print('===DUPLICATES===')
    dup_count(data)
    if check_traget:
        print(f'==={target_col}===')
        check_traget(data,target_col)
```
